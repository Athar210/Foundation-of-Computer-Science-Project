```python
import glob
import os
import pandas as pd
```

# 1. Create a single dataframe with the concatenation of all input csv files, adding a column called country


```python
files = glob.glob("trendingYT/*videos.csv.zst")

dfs = []

for file in files:
    country = os.path.basename(file)[:2] 
    temp = pd.read_csv(file, compression="zstd", encoding="latin1")
    temp["country"] = country
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)

df.head()
```

# 2. Extract all videos that have no tag.

```python
no_tag_videos = df[
    (df["tags"].isna()) |
    (df["tags"].str.strip() == "") |
    (df["tags"] == "[none]")
]

no_tag_videos.head(15)
```

# 3. For each channel, determine the total number of views

```python
channel_views = (
    df
    .groupby("channel_title", as_index=False)["views"]
    .sum()
)
channel_views.sort_values(by="views", ascending=False).head(10)
```

# 4. Save all rows with disabled comments and disabled ratings, or that have video_error_or_removed in a new dataframe called excluded, and remove those rows from the original dataframe.

```python
excluded = df[
    (df["comments_disabled"] == True) |
    (df["ratings_disabled"] == True) |
    (df["video_error_or_removed"] == True)
].copy()
```

```python
df = df.drop(excluded.index).reset_index(drop=True)
```

```python
print("Excluded rows:", excluded.shape[0])
print("Remaining rows in df:", df.shape[0])
```

# 5. Add a like_ratio column storing the ratio between the number of likes and of dislikes

```python
df["like_ratio"] = df["likes"] / df["dislikes"].replace(0, pd.NA)
df[["likes", "dislikes", "like_ratio"]].head()
```

# 6. Cluster the publish time into 10-minute intervals (e.g. from 02:20 to 02:30)

```python
df["publish_time"] = pd.to_datetime(df["publish_time"])
df["publish_interval"] = df["publish_time"].dt.floor("10min")
df[["publish_time", "publish_interval"]].head(10)
```

# 7. For each interval, determine the number of videos, average number of likes and of dislikes.

```python
interval_stats = (
    df
    .groupby("publish_interval")
    .agg(
        video_count=("video_id", "count"),
        avg_likes=("likes", "mean"),
        avg_dislikes=("dislikes", "mean")
    )
    .reset_index()
)
interval_stats.sort_values("video_count", ascending=False).head(10)
```

# 8. For each tag, determine the number of videos
### Notice that tags contains a string with several tags.


```python
df["tag_list"] = df["tags"].str.split("|")


tags_exploded = df.explode("tag_list")
tags_exploded = tags_exploded[~tags_exploded["tag_list"].isin(["[none]", ""])]


tag_counts = tags_exploded.groupby("tag_list")["video_id"].count().reset_index()


tag_counts.columns = ["tag", "video_count"]


tag_counts = tag_counts.sort_values("video_count", ascending=False)


tag_counts.head(10)
```

# 9. Find the tags with the largest number of videos

```python
top_tags = tag_counts.sort_values("video_count", ascending=False)
top_tags.head(10)
```

# 10. For each (tag, country) pair, compute average ratio likes/dislikes

```python
tags_exploded_clean = tags_exploded[~tags_exploded["tag_list"].isin(["[none]", ""])]


tag_country_ratio = (
    tags_exploded_clean
    .groupby(["tag_list", "country"], as_index=False)["like_ratio"]
    .mean()
)

tag_country_ratio.columns = ["tag", "country", "avg_like_ratio"]

tag_country_ratio = tag_country_ratio.sort_values("avg_like_ratio", ascending=False)

tag_country_ratio.head(100)
```

# 11. For each (trending_date, country) pair, the video with the largest number of views

```python
df["trending_date"] = pd.to_datetime(df["trending_date"], format="%y.%d.%m")  


top_videos_daily = (
    df.loc[df.groupby(["trending_date", "country"])["views"].idxmax()]
    .reset_index(drop=True)
)

top_videos_daily.head()
```

# 12. Divide trending_date into three columns: year, month, day

```python
df["year"] = df["trending_date"].dt.year
df["month"] = df["trending_date"].dt.month
df["day"] = df["trending_date"].dt.day

df[["trending_date", "year", "month", "day"]].head()
```

# 13. For each (month, country) pair, the video with the largest number of views


```python
idx = df.groupby(["month", "country"])["views"].idxmax()


top_videos_month = df.loc[idx]


top_videos_month = top_videos_month.sort_values(["country", "month"])


top_videos_month.head()

```

# 14. Read all json files with the video categories

```python
import json


category_files = glob.glob(r"trendingYT\*_category_id.json")  
category_list = []


for file in category_files:
    country = os.path.basename(file).split("_")[0].upper()  
    
    with open(file, encoding="utf-8") as f:
        data = json.load(f)
    

    temp_df = pd.DataFrame([{
        "category_id": int(item["id"]),
        "category_title": item["snippet"]["title"],
        "country": country
    } for item in data.get("items", [])])
    
    category_list.append(temp_df)


categories_df = pd.concat(category_list, ignore_index=True)


categories_df.head()
```

# 15. For each country, determine how many videos have a category that is not assignable.

```python

df_cat = df.merge(
    categories_df,
    on=["country", "category_id"],
    how="left"  
)


df_cat["category_title"] = df_cat["category_title"].fillna("Not assignable")


not_assignable_counts = (
    df_cat[df_cat["category_title"] == "Not assignable"]
    .groupby("country")["video_id"]
    .count()
    .to_dict()
)

not_assignable_counts
```
